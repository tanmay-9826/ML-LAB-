```python
# Step 1: Import required libraries
import pandas as pd
import numpy as np

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

# Step 2: Load the Titanic dataset
df = pd.read_csv("titanic.csv")

print("Original Dataset Shape:", df.shape)

# Step 3: Remove unnecessary columns
df = df.drop(
    columns=["PassengerId", "Name", "Ticket", "Cabin"],
    errors="ignore"
)

# Step 4: Handle missing values
if "Age" in df.columns:
    df["Age"] = df["Age"].fillna(df["Age"].median())

if "Embarked" in df.columns:
    df["Embarked"] = df["Embarked"].fillna(
        df["Embarked"].mode()[0]
    )

# Fill missing Fare values if present
if "Fare" in df.columns:
    df["Fare"] = df["Fare"].fillna(df["Fare"].median())

# Fill missing values in other selected columns, if any
for col in df.columns:
    if col != "Survived" and df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col] = df[col].fillna(df[col].median())
        else:
            mode_values = df[col].mode()
            if not mode_values.empty:
                df[col] = df[col].fillna(mode_values[0])

print("\nMissing Values After Handling:")
print(df.isnull().sum())

# Step 5: Encode categorical columns
categorical_cols = [
    col for col in ["Sex", "Embarked"]
    if col in df.columns
]

df = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)

print("\nDataset After Encoding:")
print(df.head())

# Step 6: Apply Min-Max scaling
numeric_cols = [
    col for col in ["Age", "Fare", "SibSp", "Parch"]
    if col in df.columns
]

scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

print("\nDataset After Scaling:")
print(df.head())

# Step 7: Separate features and target
X = df.drop(columns=["Survived"])
y = df["Survived"]

print("\nFeatures Shape:", X.shape)
print("Target Shape:", y.shape)

# Step 8: Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Step 9: Display final results
print("\nTraining Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)
print("Training Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)

print("\nPreprocessing Completed Successfully!")
```
