```python
# Step 1: Import the required libraries
import pandas as pd
import numpy as np

# Step 2: Load the Titanic dataset
df = pd.read_csv("titanic.csv")

# Step 3: Display the dataset
print("Titanic Dataset:")
print(df)

# Step 4: Display the first five rows
print("\nFirst five rows:")
print(df.head())

# Step 5: Display the last five rows
print("\nLast five rows:")
print(df.tail())

# Step 6: Display the number of rows and columns
print("\nDataset Shape:")
print(df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns)

# Step 7: Display dataset information
print("\nDataset Information:")
df.info()

# Step 8: Display statistical information
print("\nStatistical Description:")
print(df.describe())

# Step 9: Count missing values in each column
print("\nMissing Values:")
print(df.isnull().sum())

# Step 10: Identify numerical and categorical columns
numerical_columns = df.select_dtypes(
    include=np.number
).columns

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

print("\nNumerical Columns:")
print(numerical_columns)

print("\nCategorical Columns:")
print(categorical_columns)
```
